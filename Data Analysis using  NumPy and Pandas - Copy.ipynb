## Numpy Array Operations:

```python
import numpy as np
import pandas as pd
```

```python
# 1. Create a 1D NumPy Array:

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

```python
# 2. Inspection and Properties:

print("Shape:", temperatures_w1.shape)

print("Data Type:", temperatures_w1.dtype)

print("Number of Elements:", temperatures_w1.size)
```

```python
# 3. Array Operations

fahrenheit = (temperatures_w1 * 9/5) + 32

print("Temperature in Fahrenheit:", fahrenheit)
```

```python
print("Maximum Temperature:", np.max(temperatures_w1))

print("Minimum Temperature:", np.min(temperatures_w1))

print("Mean Temperature:", np.mean(temperatures_w1))
```

```python
# 4. Array Slicing and Indexing: 

print("first three days of weeks:",temperatures_w1[0:3])
print("for weekends:",temperatures_w1[-2:])
print("middle three days :",temperatures_w1[2:5])
```

```python
# 5. Create a 2D Array: 

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])
print(temperatures)
```

```python
# 6. Inspect and Slice the 2D Array: 

print("Shape:", temperatures.shape)

print("Data Type:", temperatures.dtype)

print("Total Elements:", temperatures.size)
```

```python
# Extract temperatures for each week.
print("Week 1:", temperatures[0])
print("Week 2:", temperatures[1])
```

```python
# Extract weekend temperatures for both weeks.
print("Week 1 Weekend:", temperatures[0, -2:])
print("Week 2 Weekend:", temperatures[1, -2:])
```

## Pandas Series:  

```python
# 1. Creating Pandas Series: 
marks = pd.Series([95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)
```

```python
# 2. Indexing and Slicing: 

# the mark of the first-rank student using int position.
print(marks.iloc[0])

# the top three ranks using loc.
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

# the mark of the third-rank student using iloc.
print("third rank:",(marks.iloc[2]))

# students who scored more than 90.
print("students marks above 90:",(marks[marks > 90]))
```

```python
#3. Manipulating Series: 

# the first-rank student's mark to 100.
marks.loc['Rank1'] = 100
print(marks)
```

```python
# Remove the last-rank student.
marks = marks.drop('Rank5')
print(marks)
```

```python
# CGPA by dividing every mark by 10.
CGPA = marks / 10

print(CGPA)
```

## Pandas DataFrame: 

```python
# 1. Creating Pandas DataFrame:

data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture',
                        'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Furniture', 'Electronics'],

    'Region': ['North', 'South', 'North', 'East', 'West',
               'North', 'East', 'West', 'South', 'North'],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)

transactions
```

```python
# Group by Region and find the average amount
transactions.groupby('Region')['Amount'].mean()
```

```python
# 3. Manipulating the DataFrame:

# Modify the 'Amount' for the transaction with TransactionID 102 to 165.

transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
transactions
```

```python
# new column 'Discount' by calculating 10% of the 'Amount'.

transactions['Discount'] = transactions['Amount'] * 0.1
transactions
```

```python
# Delete the Discount column.

transactions = transactions.drop('Discount', axis=1)
print(transactions)
```

```python
# Remove the row with TransactionID 109.

transactions = transactions.drop(
    transactions[transactions['TransactionID'] == 109].index
)
transactions
```
